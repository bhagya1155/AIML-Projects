```python
# Remove duplicate records
df.drop_duplicates(inplace=True)
```

```python
# Verify that duplicate records have been removed
print(df.duplicated().sum())
```

```python
print(df.mean(numeric_only=True))
```

```python
print(df.mode().iloc[0])
```

```python
# Calculate variance
print(df.var(numeric_only=True))
```

```python
print(df.std(numeric_only=True))
```

```python
# Calculate the correlation between numerical columns
correlation = df.corr(numeric_only=True)
print(correlation)
```

```python
# Display the correlation matrix
plt.figure(figsize=(10,8))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
# Detect outliers in Fare
plt.boxplot(df["Fare"])
plt.title("Fare Distribution")
plt.ylabel("Fare")
plt.show()
```

```python
# Plot survival count
df["Survived"].value_counts().plot(kind="bar")
plt.title("Passenger Survival")
plt.xlabel("Survived")
plt.ylabel("Count")
plt.show()
```

```python
# Plot age distribution
plt.hist(df["Age"].dropna(), bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Plot relationship between age and fare
plt.scatter(df["Age"], df["Fare"])
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
# Filter passengers older than 30 years
age_filter = df[df["Age"] > 30]
print(age_filter.head())
```

```python
# Filter female passengers
female = df[df["Sex"] == "female"]
print(female.head())
```

```python
# Filter first-class passengers
first_class = df[df["Pclass"] == 1]
print(first_class.head())
```

```python
# Calculate average fare for each passenger class
group = df.groupby("Pclass")["Fare"].mean()
print(group)
```

```python
# Count survived and non-survived passengers
print(df["Survived"].value_counts())
```
