```python
plt.figure(figsize=(6,4))
plt.boxplot(df["overall_score"])
plt.title("Box Plot of Exam Score")
plt.ylabel("Exam Score")
plt.show()
```

```python
# Display the distribution of overall scores
df["overall_score"].value_counts().plot(kind="bar")
plt.title("Overall Score Distribution")
plt.xlabel("Overall Score")
plt.ylabel("Count")
plt.show()
```

```python
# Display the distribution of overall scores
plt.hist(df["overall_score"], bins=10)
plt.title("Overall Score Distribution")
plt.xlabel("Overall Score")
plt.ylabel("Frequency")
plt.show()
```

```python
# Display the trend of overall scores
plt.plot(df["overall_score"])
plt.title("Overall Score")
plt.xlabel("Students")
plt.ylabel("Overall Score")
plt.show()
```

```python
# Analyze the relationship between study hours and overall score
plt.scatter(df["study_hours"], df["overall_score"])
plt.title("Study Hours vs Overall Score")
plt.xlabel("Study Hours")
plt.ylabel("Overall Score")
plt.show()
```
